```python
import lightkurve as lk
import matplotlib.pyplot as plt
```

```python
search = lk.search_lightcurve("WASP-96", mission="TESS", author="SPOC", exptime=120)
print(f"Našiel som {len(search)} súborov")
lc = search[0].download() # beriem len jeden sektor, stačí
lc = lc.remove_nans()
lc.plot()
plt.show()
```

```python
flat = lc.remove_outliers(sigma=5).flatten(window_length=501)

# WASP-96 b má periódu 3.425 dní, hľadáme okolo toho
periods = flat.to_periodogram(method='bls', period=[3, 4], duration=0.2)
print(f"Našiel som: {periods.period_at_max_power.value:.3f} dní (reálne je 3.425)")

# poskladaj to
folded = flat.fold(period=periods.period_at_max_power)
folded.plot()
plt.title("WASP-96 b - poskladaný tranzit")
plt.show()

periods.plot()
plt.show()
```
