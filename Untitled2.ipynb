```python
import lightkurve as lk
import matplotlib.pyplot as plt

TARGET = "TOI-4600"
search = lk.search_lightcurve(TARGET)
print(f"Nasiel {len(search)} suborov, stahujem prvych 20...")

# stahujeme po jednom, nie download_all - to fixne tu chybu
lcs = []
for s in search[:20]:
    try:
        lc = s.download()
        if lc is not None:
            lcs.append(lc)
    except:
        pass

print(f"Stiahlo sa {len(lcs)} kusov, lepim...")

# zlepime
from lightkurve.collections import LightCurveCollection
collection = LightCurveCollection(lcs)
lc_big = collection.stitch().remove_nans().flatten(window_length=401)

plt.figure(figsize=(18,4))
lc_big.plot()
plt.title(f"{TARGET} - 20 sektorov - HLADAJ JEDEN VELKY ZUB")
plt.show()

print("Hotovo! Vidis tam jednu velku dieru?")
```

```python
import lightkurve as lk
import matplotlib.pyplot as plt
import numpy as np
from astropy.timeseries import BoxLeastSquares

TARGET = "TOI-1859"
search = lk.search_lightcurve(TARGET)

# LEN 2 SUBORY - RYCHLE
lc = search[0].download().remove_nans().flatten(401)
lc2 = search[1].download().remove_nans().flatten(401)
from lightkurve.collections import LightCurveCollection
lc_flat = LightCurveCollection([lc, lc2]).stitch()

model = BoxLeastSquares(lc_flat.time.value, lc_flat.flux.value)
durations = [0.1, 0.2, 0.3]
result = model.autopower(durations, objective='snr')
result = result[(result.period > 1) & (result.period < 8)]
best = result.period[np.argmax(result.power)]

print(f"PERIODA: {best:.4f} dni")

lc_flat.fold(period=best).scatter(s=3, c='black')
plt.title(f"{TARGET} - {best:.3f} dni")
plt.show()
```
